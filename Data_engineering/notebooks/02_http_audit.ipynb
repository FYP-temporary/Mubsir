```python
from pathlib import Path
import sys

import duckdb


# ============================================================
# PROJECT CONFIG
# ============================================================

current = Path.cwd().resolve()

PROJECT_ROOT = next(
    p for p in [current, *current.parents]
    if (p / "src" / "data_pipeline" / "config.py").exists()
)

sys.path.insert(
    0,
    str(PROJECT_ROOT / "src" / "data_pipeline")
)

from config import HTTP_CSV


# ============================================================
# HTTP DATA AUDIT
# ============================================================

http_path = HTTP_CSV.resolve().as_posix()

con = duckdb.connect()

print("=" * 70)
print("HTTP DATA AUDIT")
print("=" * 70)

print("\nHTTP schema:")

schema = con.execute(f"""
    DESCRIBE
    SELECT *
    FROM read_csv_auto(
        '{http_path}',
        types={{'date': 'VARCHAR'}}
    )
""").df()

display(schema)


print("\nSample records:")

sample = con.execute(f"""
    SELECT
        id,
        date,
        user,
        pc,
        url
    FROM read_csv_auto(
        '{http_path}',
        types={{'date': 'VARCHAR'}}
    )
    LIMIT 5
""").df()

display(sample)

con.close()

print("\n✅ HTTP AUDIT COMPLETE")
```
